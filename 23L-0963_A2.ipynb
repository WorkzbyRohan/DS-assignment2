```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from collections import Counter

df = pd.read_csv('netflix_titles.csv')

print("Shape:", df.shape)
print()
print("Sample records:")
df.head(5)
```

```python

print("Director nulls by type:")
print(df[df['director'].isnull()]['type'].value_counts())
print()
print("Cast nulls by type:")
print(df[df['cast'].isnull()]['type'].value_counts())
```

```python

print("Full-row duplicates:", df.duplicated().sum())
print("Duplicate show_id:", df['show_id'].duplicated().sum())
print()

dup_title = df[df.duplicated(subset=['title','type'], keep=False)]
print("Duplicate (title + type) pairs:", len(dup_title))
if len(dup_title) > 0:
    print(dup_title[['show_id','type','title','release_year']].head(10))
```

```python

print("date_added dtype:", df['date_added'].dtype, "-> should be datetime")
print("release_year dtype:", df['release_year'].dtype, "-> OK (int64)")
print()
print("Sample date_added values:", df['date_added'].dropna().head(3).tolist())
```

```python

print("All unique rating values:")
print(df['rating'].value_counts(dropna=False))
```

```python

bad_rating = df[df['rating'].isin(['74 min', '84 min', '66 min'])]
print("Rows with duration in rating column:")
print(bad_rating[['show_id','title','type','rating','duration']])
```

```python

print("release_year range:", df['release_year'].min(), "to", df['release_year'].max())
print()
print("Titles released before 1970:")
print(df[df['release_year'] < 1970][['title','release_year','type']].sort_values('release_year').head(10))
```

```python

Q1 = df['release_year'].quantile(0.25)
Q3 = df['release_year'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_year = df[(df['release_year'] < lower) | (df['release_year'] > upper)]
print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}")
print(f"Lower fence: {lower}, Upper fence: {upper}")
print(f"Outlier count: {len(outliers_year)}")
print(outliers_year[['title','release_year','type']].sort_values('release_year').head(8))
```

```python

print("Max countries per title:", df['country'].dropna().str.split(',').apply(len).max())
print("Max genres per title:", df['listed_in'].str.split(',').apply(len).max())
print("Max cast per title:", df['cast'].dropna().str.split(',').apply(len).max())
```

```python

netflix = df.copy()


bad_idx = netflix[netflix['rating'].isin(['74 min', '84 min', '66 min'])].index
for idx in bad_idx:
    netflix.loc[idx, 'duration'] = netflix.loc[idx, 'rating']
    netflix.loc[idx, 'rating'] = np.nan

print("After fix — bad rating rows:")
print(netflix.loc[bad_idx, ['title','rating','duration']])
```

```python

netflix['date_added'] = pd.to_datetime(
    netflix['date_added'].str.strip(), format='%B %d, %Y', errors='coerce'
)
print("date_added dtype:", netflix['date_added'].dtype)
print("NaT count:", netflix['date_added'].isnull().sum())
```

```python

netflix['director'] = netflix['director'].fillna('Unknown')
netflix['cast'] = netflix['cast'].fillna('Unknown')
netflix['country'] = netflix['country'].fillna('Unknown')

print("Nulls after fill:")
print(netflix[['director','cast','country']].isnull().sum())
```

```python



netflix['year_added'] = netflix['date_added'].dt.year
netflix['month_added'] = netflix['date_added'].dt.month


netflix['duration_int'] = netflix['duration'].str.extract(r'(\d+)').astype(float)


netflix['duration_unit'] = netflix['duration'].str.replace(r'\d+\s*', '', regex=True).str.strip()


netflix['num_countries'] = netflix['country'].apply(
    lambda x: len(x.split(',')) if x != 'Unknown' else 0
)


netflix['num_genres'] = netflix['listed_in'].apply(lambda x: len(x.split(',')))


netflix['gap_years'] = netflix['year_added'] - netflix['release_year']

print("Derived columns sample:")
print(netflix[['year_added','month_added','duration_int','duration_unit',
               'num_countries','num_genres','gap_years']].head())
```

```python

print("Final missing value counts:")
print(netflix.isnull().sum())
print()
print("Final shape:", netflix.shape)
```

```python

netflix.to_csv('23L-0614_A2_netflix_cleaned.csv', index=False)
print("Saved: 23L-0614_A2_netflix_cleaned.csv")
```

```python
valid_ratings = netflix['rating'].dropna()
rating_counts = valid_ratings.value_counts()
print("Rating counts:")
print(rating_counts)

fig, ax = plt.subplots(figsize=(10,4))
rating_counts.plot(kind='bar', ax=ax, color='#E50914', edgecolor='black')
ax.set_title('Maturity Rating Distribution', fontsize=13)
ax.set_xlabel('Rating')
ax.set_ylabel('Count')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig('a4_rating.png', dpi=150)
plt.show()
```

```python
movies = netflix[netflix['type'] == 'Movie'].copy()
print("Movie Duration (minutes) Statistics:")
print(movies['duration_int'].describe())
print(f"Skewness: {movies['duration_int'].skew():.3f}")

fig, ax = plt.subplots(figsize=(10,4))
ax.hist(movies['duration_int'].dropna(), bins=40, color='steelblue', edgecolor='black', alpha=0.85)
ax.axvline(movies['duration_int'].median(), color='red', linestyle='--',
           label=f"Median: {movies['duration_int'].median():.0f} min")
ax.axvline(movies['duration_int'].mean(), color='orange', linestyle='--',
           label=f"Mean: {movies['duration_int'].mean():.1f} min")
ax.set_title('Distribution of Movie Durations (minutes)', fontsize=13)
ax.set_xlabel('Duration (minutes)')
ax.set_ylabel('Number of Movies')
ax.legend()
plt.tight_layout()
plt.savefig('a4_movie_duration.png', dpi=150)
plt.show()


Q1 = movies['duration_int'].quantile(0.25)
Q3 = movies['duration_int'].quantile(0.75)
IQR = Q3 - Q1
print(f"\nIQR fences: lower={Q1-1.5*IQR:.0f} min, upper={Q3+1.5*IQR:.0f} min")
print("Short outliers (<lower fence):", (movies['duration_int'] < Q1-1.5*IQR).sum())
print("Long outliers (>upper fence):", (movies['duration_int'] > Q3+1.5*IQR).sum())
```

**Interpretation:** Movie durations follow a roughly normal distribution centred around 98 minutes (the global standard for feature films). The slight right skew comes from a handful of very long films (200+ min). Short outliers (under ~52 min) include short films and stand-up specials. Long outliers are genuine epic films. All are retained as real content.

```python

all_countries = []
for entry in netflix['country']:
    if entry != 'Unknown':
        for c in entry.split(','):
            all_countries.append(c.strip())

country_counts = Counter(all_countries)
top_countries = pd.DataFrame(country_counts.most_common(15), columns=['Country','Count'])

fig, ax = plt.subplots(figsize=(10,5))
ax.barh(top_countries['Country'][::-1], top_countries['Count'][::-1],
        color='#E50914', edgecolor='black')
ax.set_title('Top 15 Countries by Number of Titles', fontsize=13)
ax.set_xlabel('Number of Titles')
plt.tight_layout()
plt.savefig('a4_countries.png', dpi=150)
plt.show()
print(top_countries)
```

```python

all_genres = []
for entry in netflix['listed_in']:
    for g in entry.split(','):
        all_genres.append(g.strip())

genre_counts = Counter(all_genres)
top_genres = pd.DataFrame(genre_counts.most_common(15), columns=['Genre','Count'])

fig, ax = plt.subplots(figsize=(10,5))
ax.barh(top_genres['Genre'][::-1], top_genres['Count'][::-1],
        color='steelblue', edgecolor='black')
ax.set_title('Top 15 Genres on Netflix', fontsize=13)
ax.set_xlabel('Number of Titles')
plt.tight_layout()
plt.savefig('a4_genres.png', dpi=150)
plt.show()
print(top_genres)
```

**Interpretation:** The US leads by a huge margin, followed by India and the UK. For genres, International Movies and Dramas are the most common. The "International Movies" tag is a catch-all for non-US content, which inflates its count — it is not a true genre like Action or Comedy. This must be kept in mind for any genre-based analysis.


```python
movie_rated = netflix[(netflix['type']=='Movie') &
                       netflix['rating'].notna() &
                       ~netflix['rating'].isin(['NR','UR'])].copy()

order = movie_rated.groupby('rating')['duration_int'].median().sort_values().index

fig, ax = plt.subplots(figsize=(12,5))
sns.boxplot(data=movie_rated, x='rating', y='duration_int', order=order,
            ax=ax, palette='Set2')
ax.set_title('Movie Duration (min) by Maturity Rating', fontsize=13)
ax.set_xlabel('Rating')
ax.set_ylabel('Duration (minutes)')
plt.tight_layout()
plt.savefig('a5_duration_rating.png', dpi=150)
plt.show()

print("Median movie duration by rating:")
print(movie_rated.groupby('rating')['duration_int'].median().sort_values().round(1))
```

**Interpretation:** There is a weak relationship — NC-17 and R-rated movies tend to run slightly longer, while G-rated movies are shortest. This is consistent with the general pattern where children's films are shorter and adult dramas/thrillers tend to run longer. However, the distributions overlap heavily, meaning rating alone is not a strong predictor of duration.

**Interpretation:** There is a moderate positive correlation. Most titles cluster near the diagonal (released and added around the same period), but many older titles sit far below it — released decades ago, added recently. This confirms Netflix's catalogue has two components: fresh content and older library acquisitions. The wide vertical spread of older release years shows that Netflix aggressively added library content in 2016-2020.


```python
ct = pd.crosstab(netflix['type'], netflix['rating'])
major_ratings = ['G','PG','PG-13','R','NC-17','TV-Y','TV-Y7','TV-Y7-FV','TV-G','TV-PG','TV-14','TV-MA']
ct = ct[[c for c in major_ratings if c in ct.columns]]
ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(13,5))
ct_pct.plot(kind='bar', ax=ax, colormap='tab20', edgecolor='black', width=0.6)
ax.set_title('Rating Distribution by Content Type (% within each type)', fontsize=13)
ax.set_xlabel('Content Type')
ax.set_ylabel('Percentage (%)')
ax.tick_params(axis='x', rotation=0)
ax.legend(title='Rating', bbox_to_anchor=(1,1), loc='upper left')
plt.tight_layout()
plt.savefig('a5_type_rating.png', dpi=150)
plt.show()

print("Rating % by content type (rounded):")
print(ct_pct.round(1))
```

**Interpretation:** Movies use MPAA ratings (G, PG, PG-13, R, NC-17) while TV Shows use the TV Parental Guidelines (TV-Y through TV-MA). This division is consistent and expected. Among Movies, TV-MA and R are most common. Among TV Shows, TV-MA makes up ~40% of all TV Show ratings. Both content types skew heavily towards mature audiences on Netflix.


```python
type_year = (netflix.dropna(subset=['year_added'])
              .groupby(['year_added','type'])
              .size()
              .unstack(fill_value=0))
type_year = type_year[type_year.index >= 2015]

fig, axes = plt.subplots(1, 2, figsize=(14,5))


type_year.plot(kind='bar', stacked=True, ax=axes[0],
               color=['#E50914','steelblue'], edgecolor='black', width=0.7)
axes[0].set_title('Titles Added per Year by Type', fontsize=12)
axes[0].set_xlabel('Year Added')
axes[0].set_ylabel('Number of Titles')
axes[0].tick_params(axis='x', rotation=45)
axes[0].legend(title='Type')


tv_share = type_year['TV Show'] / type_year.sum(axis=1) * 100
axes[1].plot(tv_share.index, tv_share.values, marker='o', color='steelblue', linewidth=2)
axes[1].set_title('TV Show Share (%) by Year Added', fontsize=12)
axes[1].set_xlabel('Year Added')
axes[1].set_ylabel('TV Show Share (%)')
axes[1].set_ylim(0, 60)
axes[1].axhline(30, color='gray', linestyle='--', alpha=0.5, label='30% line')
axes[1].legend()

plt.tight_layout()
plt.savefig('a6_type_year_multi.png', dpi=150)
plt.show()

print("TV Show % by year added:")
print(tv_share.round(1))
```

**Interpretation:** Adding type as a third dimension to the year-addition view reveals the trend clearly. The TV Show share grew from around 22-25% in 2015-2016 to above 35% in 2019-2020. Total additions peaked around 2019-2020 and dropped sharply in 2021 (likely because the dataset's cutoff is late 2021, not because additions actually stopped). This growth in TV Show share is consistent with Netflix's known shift towards original and acquired series.


```python
dated = netflix.dropna(subset=['year_added','month_added'])
dated = dated[dated['year_added'] >= 2016].copy()

month_names = {1:'Jan',2:'Feb',3:'Mar',4:'Apr',5:'May',6:'Jun',
               7:'Jul',8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}

fig, axes = plt.subplots(2, 1, figsize=(13,8))

for i, content_type in enumerate(['Movie','TV Show']):
    subset = dated[dated['type']==content_type]
    hm = subset.groupby(['year_added','month_added']).size().unstack(fill_value=0)
    hm.columns = [month_names[c] for c in hm.columns]
    cmap = 'Reds' if content_type == 'Movie' else 'Blues'
    sns.heatmap(hm, ax=axes[i], cmap=cmap, annot=True, fmt='d',
                linewidths=0.3, cbar_kws={'label':'Count'})
    axes[i].set_title(f'{content_type}s Added per Month-Year', fontsize=12)
    axes[i].set_xlabel('Month')
    axes[i].set_ylabel('Year')

plt.tight_layout()
plt.savefig('a6_heatmap.png', dpi=150)
plt.show()
```

**Interpretation:** The heatmap adds month as a third variable. Both content types show bursts of additions in Q4 (October-December), likely aligning with holiday viewing season. 2019 and early 2020 are the most active periods for both types. The 2021 column is sparse because the dataset likely covers only part of that year. No single month dominates consistently, but the Q4 pattern appears across multiple years.


### (3) Gap Years by Country and Type
**Question:** Does the release-to-Netflix gap differ by country and content type?


```python
netflix['primary_country'] = netflix['country'].apply(
    lambda x: x.split(',')[0].strip() if x != 'Unknown' else 'Unknown'
)

top5 = ['United States','India','United Kingdom','Japan','South Korea']
gap_subset = netflix[
    netflix['primary_country'].isin(top5) &
    netflix['gap_years'].notna() &
    (netflix['gap_years'] >= 0) &
    (netflix['gap_years'] <= 50)
]

fig, ax = plt.subplots(figsize=(12,5))
sns.boxplot(data=gap_subset, x='primary_country', y='gap_years', hue='type',
            palette={'Movie':'#E50914','TV Show':'steelblue'}, ax=ax)
ax.set_title('Release-to-Netflix Gap by Country and Content Type', fontsize=13)
ax.set_xlabel('Primary Country')
ax.set_ylabel('Gap (Years)')
ax.legend(title='Type')
plt.tight_layout()
plt.savefig('a6_gap_country_type.png', dpi=150)
plt.show()

print("Median gap by country and type:")
print(gap_subset.groupby(['primary_country','type'])['gap_years'].median().unstack().round(1))
```

**Interpretation:** This three-variable view reveals clear country-level differences. South Korean TV Shows have a very small median gap (near 0-1 year), consistent with Netflix actively licensing K-dramas as they air. Indian Movies have a wide distribution — some are added immediately (Bollywood originals or licensed releases) while others are catalogue acquisitions from years ago. US Movies have a moderate gap, and the wide spread reflects the large variety (new releases mixed with old library films). Japan's TV Shows (mostly anime) also tend to have a small gap, reflecting Netflix's licensing of seasonal anime.


```python

type_release = (netflix[netflix['release_year'] >= 2005]
                .groupby(['release_year','type']).size().unstack(fill_value=0))

type_added = (netflix.dropna(subset=['year_added'])
              .groupby(['year_added','type']).size().unstack(fill_value=0))
type_added = type_added[type_added.index >= 2014]

fig, axes = plt.subplots(1, 2, figsize=(14,5))


axes[0].stackplot(type_release.index,
                  type_release['Movie'], type_release.get('TV Show', 0),
                  labels=['Movie','TV Show'], colors=['#E50914','steelblue'], alpha=0.85)
axes[0].set_title('Content Mix by Release Year (2005+)', fontsize=12)
axes[0].set_xlabel('Release Year')
axes[0].set_ylabel('Number of Titles')
axes[0].legend(loc='upper left')


axes[1].stackplot(type_added.index,
                  type_added['Movie'], type_added.get('TV Show', 0),
                  labels=['Movie','TV Show'], colors=['#E50914','steelblue'], alpha=0.85)
axes[1].set_title('Content Mix by Year Added to Netflix', fontsize=12)
axes[1].set_xlabel('Year Added')
axes[1].set_ylabel('Number of Titles')
axes[1].legend(loc='upper left')

plt.tight_layout()
plt.savefig('a7_q1_mix.png', dpi=150)
plt.show()

print("TV Show % by year added:")
tv_pct = type_added['TV Show'] / type_added.sum(axis=1) * 100
print(tv_pct.round(1))
```

**Interpretation:** By year_added (the more reliable indicator of Netflix strategy), the TV Show share grew from ~22% in 2016 to ~35-38% by 2019-2020. Looking at release year, the TV Show share among newer releases is also higher than for older titles. Both views confirm a clear trend: Netflix has been adding proportionally more TV Shows over time, consistent with its known push into original and acquired series content.


```python
gap_clean = netflix[(netflix['gap_years'] >= 0) & (netflix['gap_years'] <= 70)].copy()

print("Gap statistics:")
print(gap_clean['gap_years'].describe())
print(f"\nTitles with gap = 0 (same year): {(gap_clean['gap_years']==0).sum()}")
print(f"Titles with gap > 20 years: {(gap_clean['gap_years']>20).sum()}")

fig, axes = plt.subplots(1, 2, figsize=(13,5))

axes[0].hist(gap_clean['gap_years'], bins=40, color='#E50914', edgecolor='black', alpha=0.85)
axes[0].axvline(gap_clean['gap_years'].median(), color='black', linestyle='--',
                label=f"Median: {gap_clean['gap_years'].median():.0f} yr")
axes[0].set_title('Release-to-Netflix Gap Distribution', fontsize=12)
axes[0].set_xlabel('Gap (Years)')
axes[0].set_ylabel('Number of Titles')
axes[0].legend()

sns.boxplot(data=gap_clean, x='type', y='gap_years',
            palette={'Movie':'#E50914','TV Show':'steelblue'}, ax=axes[1])
axes[1].set_title('Gap by Content Type', fontsize=12)
axes[1].set_xlabel('Content Type')
axes[1].set_ylabel('Gap (Years)')

plt.tight_layout()
plt.savefig('a7_q2_gap.png', dpi=150)
plt.show()

print("\nMedian gap by type:")
print(gap_clean.groupby('type')['gap_years'].median())
```

**Interpretation:** The median gap is around 1-2 years, showing most content is added relatively soon after release. The right-skewed distribution reveals a long tail of old library content. Movies have a wider gap distribution than TV Shows — Netflix picks up old movies more aggressively than old TV Shows. Titles with gap=0 are likely Netflix Originals or day-and-date licensing deals where content arrives on Netflix the same year it is produced.


```python

exploded = netflix[netflix['country'] != 'Unknown'].copy()
exploded = exploded.assign(csplit=exploded['country'].str.split(',')).explode('csplit')
exploded['csplit'] = exploded['csplit'].str.strip()
exploded = exploded.assign(gsplit=exploded['listed_in'].str.split(',')).explode('gsplit')
exploded['gsplit'] = exploded['gsplit'].str.strip()

top5 = ['United States','India','United Kingdom','Japan','South Korea']
cg = (exploded[exploded['csplit'].isin(top5)]
      .groupby(['csplit','gsplit']).size().reset_index(name='count'))

fig, axes = plt.subplots(1, 5, figsize=(20,6))
for i, country in enumerate(top5):
    data = cg[cg['csplit']==country].nlargest(6,'count').sort_values('count')
    axes[i].barh(data['gsplit'], data['count'], color='#E50914', edgecolor='black')
    axes[i].set_title(country, fontsize=9)
    axes[i].set_xlabel('Count')
    axes[i].tick_params(axis='y', labelsize=7)

plt.suptitle('Top 6 Genres per Top-5 Country', fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig('a7_q3_genre_country.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Interpretation:** Genre profiles differ significantly by country. The US has a wide spread — Dramas, Documentaries, Stand-Up Comedy, and Thrillers are all prominent. India is dominated by International Movies and Dramas, reflecting Bollywood's output. Japan's profile is unique — Anime is a top genre that barely appears elsewhere. South Korea's top genres are International TV Shows and Romantic TV Shows (K-dramas). The UK leans towards British TV Shows and Docuseries. This means a genre-level analysis without country context can be misleading.

```python
rating_col = 'Customer stratification rating'

fig, axes = plt.subplots(1, 3, figsize=(15, 5))


axes[0].scatter(smarket['Unit price'], smarket[rating_col],
                alpha=0.2, color='steelblue', s=12)
r1, p1 = stats.pearsonr(smarket['Unit price'], smarket[rating_col])
axes[0].set_title(f'Rating vs Unit Pricer={r1:.3f}, p={p1:.3f}', fontsize=11)
axes[0].set_xlabel('Unit Price')
axes[0].set_ylabel('Rating')


axes[1].scatter(smarket['Total'], smarket[rating_col],
                alpha=0.2, color='#E50914', s=12)
r2, p2 = stats.pearsonr(smarket['Total'], smarket[rating_col])
axes[1].set_title(f'Rating vs Totalr={r2:.3f}, p={p2:.3f}', fontsize=11)
axes[1].set_xlabel('Total ($)')
axes[1].set_ylabel('Rating')


sns.boxplot(data=smarket, x='Product line', y=rating_col, palette='Set2', ax=axes[2])
axes[2].set_title('Rating by Product Line', fontsize=11)
axes[2].set_xlabel('Product Line')
axes[2].set_ylabel('Rating')
axes[2].tick_params(axis='x', rotation=25)

plt.suptitle('Customer Rating vs Key Variables', fontsize=13, y=1.01)
plt.tight_layout()
plt.savefig('b7_q5_rating.png', dpi=150)
plt.show()

print("Pearson r — Rating vs Unit price:", round(r1,3), "  p:", round(p1,3))
print("Pearson r — Rating vs Total:", round(r2,3), "  p:", round(p2,3))
print()
print("Mean rating by product line:")
print(smarket.groupby('Product line')[rating_col].mean().sort_values(ascending=False).round(3))
print()
print("Mean rating by branch:")
print(smarket.groupby('Branch')[rating_col].mean().round(3))
```

**Interpretation:** Customer rating has near-zero correlation with both unit price (r≈0.01) and total spend (r≈0.01). The p-values confirm these correlations are not statistically significant. Rating by product line also shows nearly identical medians and overlapping distributions across all six product lines. The same holds for branch. This is an important null result: customer satisfaction in this dataset is essentially independent of what was bought, how much was spent, or where the purchase was made. This could mean (a) ratings reflect personal customer dispositions rather than transaction-specific satisfaction, (b) the data is synthetic and ratings were assigned independently, or (c) other unmeasured factors (staff behaviour, wait time) drive satisfaction. Regardless, `Customer stratification rating` should not be expected to be predictable from the other features in this dataset.
